#### Indexing in pandas
Indexing operator and attribute selection is easy to pick up and use. But, pandas has it's own accessor operators,`loc` and `iloc`, which can be used for more advanced operations.<br>
pandas indexing has 2 paradigms:
1. Index-based selection
2. Label-based selection

**Index-based selection**
<br>Selecting the data based on it's position in the dataset.
<br>If you wish to get the first row of the DataFrame, do the following:

````python
data.iloc[0]
````

`iloc` and `loc` are row first, column second. But this is opposite of when we used basic indexing operator, which is column first, row-second.<br>
This means that fetching rows is easier than fetching the columns.<br>
To get the first column, we do the following:

````python
data.iloc[:,0]
````

`data.iloc[:,0]` fetches all the rows and 1st column. The common syntax would be `data.iloc[row,column]`.<br>
You can specify the range of rows and columns you wish to fetch. For example, you wish to fetch the first 10 rows and 1st 4 columns:

````python
data.iloc[0:10,0:4]
````

You can notice that though we want 10 rows and 4 columns, the element at the 10th index and 4th row is not retrieved. This is because the final range is not included but the initial range is included and the index starts from 0 for both row and column.

You can also pass list:

````python
#here [0,1,3,6] represents the index of the rows
data.iloc[[0,1,3,6],0]
````

You can also use negative numbers to represent the ranges. It will start counting forward from the end ofthe values. <br>
To fetch the last 5 elements of the dataset:

````python
data.iloc[-5:]
````

**Lable-based selection**<br>
Here we use the data index value as attribute, not the position. For this, we use `loc` operator.

````python
#fetch the data present at the 0th row and the column named "customer_id".
data.loc[0,"customer_id"]
````

````python
data.loc[0:3,["customer_id","customer_name"]]
````

Notice that you can use position for rows but not or columns. The other important thing is that the final range is also included in the result. 

**Choosing between `iloc` and `loc`** <br>
You might have noticed that the way indexing works using `iloc` and `loc` is different. `iloc` uses the python's stdlib indexing, where the first elemnt is included and the last elemnt is excluded while mentioning the range.So, `0:10` will give entries `0,1,2...9` but the 10th entry is excluded.<br>
As for `loc`, 0:10 will give the entries `0,1,2,3...10`.Here, the last element is included.

#### Conditional selection 
We can ask questions based on conditions. Let us say we want to knwo whether the `customer_acquisition_cost` is morethan 13 or not for each row:



````python
data.customer_acquisition_cost >13
````

As you can see, it returns True or False based on the condition we asked.Now, you can use this inside the `loc` operator to fetch the data which return True.

````python
data.loc[data.customer_acquisition_cost>13]
````

We can aslo use 2 different condtions at the same time. For example, retrieve the data where the customer is Female and the customer_acquisition_cost is greater than 20. For this use the `&` operator.

````python
data.loc[(data.gender == "Female") & (data.customer_acquisition_cost > 20)]
````

THe customer is Female or the customer_acquisition_cost is greater than 20. Here we use `|` operator.

````python
data.loc[(data.gender =="Female") | (data.customer_acquisition_cost > 20)]
````

Pandas comes with a few built-in conditional selectors, here we will mainly deal with 2.
1. `isin()`
2. `isnull()`

`isin()` lets you select data whose value 'is in' a list of values.For example, display only those elements where the customer is either from USA or Canada.

````python
data.loc[data.customer_country.isin(["USA","Canada"])]
````

`isnull()` and it's comanion `notnull()`. These are used to retrieve any values which are null(not  ull) empty(NaN). For example, display the data where the age of cutsomer is given. 

````python
data.loc[data.customer_age.notnull()]
````

*Note*<br>
The dataset used here is a clean dataset.Hence, use the dataset with null values in it, if you wish to check the presence of the null.

##### Maps
In data science, we often need to transform the data present in one form to another, or creating new representations from the existing data. For these, we use `maps`.  There are 2 maping methods. Let us use `map()` first.

````python
data1 = data.customer_age.mean()
data.customer_age.map(lambda x: x-data1)
````

The function you pass to `map()` should expect a single value from the Series (a point value, in the above example), and return a transformed version of that value. `map()` returns a new Series where all the values have been transformed by your function.

`apply()` is another equivalent method. It is used to change the whole DataFrame by calling a custom method on each rows.

````python
def change(row):
    row.customer_age = row.customer_age - data1
    return row

data.apply(change, axis="columns")
````

If we had called `apply()` with `axis='index'`, then instead of passing a function to transform each row, we would need to give a function to transform each column.

Note that `map()` and `apply()` return new, transformed Series and DataFrames, respectively. They don't modify the original data they're called on. 

Pandas provides many common iperations as built-ins.

````python
data_mean = data.customer_age.mean()
data.customer_age - data_mean
````

Pandas will also undestand what to do if we perform these operations between series of equal length.

````python
data.customer_city + ' - ' + data.customer_state
````

### Grouping and Sorting

We have used map till now to change the data in the DataFrame or Series one at a time for entire column. Sometimes, we want to group the data and perform some operations. We will do it using `groupby()` as follws:


````python
data.groupby("customer_age").gender.value_counts()
````

We grouped the customers based on their age and then for each group, we are finding the unique list of gender and counts. 

You can think of each group as a slice of the DataFrame and then you can access them using `apply()`.

````python
data.groupby('customer_age').apply(lambda df: df.customer_age)
````

````python
data.groupby(['customer_country','region']).head()
````

Another `groupby()` method worth mentioning is `agg()` which lets us run a bunch of different functions on the DataFrame simultaneously. For example, we can generate a simple summer of the dataset.

````python
data.groupby('customer_country').agg([len])
````

#### Missing Data
Entries missing values is given the value `NaN`, short for `not a number`. For technical reasons, the data type of `NaN` is `float64`.<br>
Pandas provides methods to check for the presence of `NaN` in the column of the dataset. You can use `isnull()` to fetch the NaN values.

````python
data[pd.isnull(data.customer_age)]
````

Similarly, you can use `notnull()` to fetch all the present values. 

````python
data[pd.notnull(data.customer_age)]
````

You can also replace the null values with anything using `fillna()`.

````python
data.customer_age.fillna("Unknown")
````

Alternatively, we can replace the present non null value with any other value using `replace()` method.

````python
data.gender.replace("Male","Female")
````
