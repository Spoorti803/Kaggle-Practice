# Pandas Library 
Pandas is the most popular library of python used for data analysis.
To work with it, first import the library.(you can use alias for simplicity)

```python
import pandas as pd
```

Two main concepts in pandas are:

1. DataFrame
2. Series

We will create our own data using these as follows

**DataFrame** <br>
DataFrame is a table. It contains an array of entries. Each entry corresponds to row and coloumn.
For example: 31 is a entry whose row=0, column = yes

```python
data = pd.DataFrame({'yes':[31,3],'no':[33,9]})
print(data)
```

As you can see, the table is automatically given index values which starts for 0.
If you wish to use give specific values to index, do  follows:

```python
data = pd.DataFrame({'Bob':["I likes it","Bland"],
                     'Alice':["It was awful","I liked it"]},
                     index = ['Product A','Product B'])
print(data)
```

**Series**:
Series is a sequence of data values. If DataFrame is Table, series is a List.

```python
data = pd.Series([1,2,3,4])
data
```

Series is just a colum of DataFrame. So, you can assign a row lable using `index` parameter. However, a Series does not have a column name. It only has one name overall.

```python
data = pd.Series([30,40,50],index=['Subject A','Subject B','Subject c'],name = 'Marks of a Student')
data
```
